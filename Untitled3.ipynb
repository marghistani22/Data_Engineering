```python
import pandas as pd
import sqlite3
import matplotlib.pyplot as plt

# Create simple food dataset
data = {
    "food_name": ["Apple","Banana","Chicken","Rice","Milk","Egg","Bread","Broccoli","Salmon","Yogurt"],
    "category": ["Fruit","Fruit","Protein","Grain","Dairy","Protein","Grain","Vegetable","Protein","Dairy"],
    "calories": [52,89,165,130,42,155,265,34,208,59],
    "protein": [0.3,1.1,31,2.7,3.4,13,9,2.8,20,10]
}

df = pd.DataFrame(data)

df
```

## Create Food Dataset

We create a small dataset containing nutritional information about different foods.

The dataset includes:

- Food name
- Food category
- Calories
- Protein content

This dataset will be used to build the food database.

## Create SQLite Database

Next, we create a SQLite database and store the food dataset as a table called **foods**.

SQLite is a lightweight database system that allows us to store and query structured data.

```python
conn = sqlite3.connect("food_database.db")

df.to_sql("foods", conn, if_exists="replace", index=False)

print("Database created")
```

```python
query = """
SELECT food_name, calories
FROM foods
ORDER BY calories DESC
LIMIT 5
"""

top_foods = pd.read_sql_query(query, conn)

top_foods
```

```python
query_top_calories = """
SELECT food_name, category, calories
FROM foods
ORDER BY calories DESC
LIMIT 5
"""

top_calories = pd.read_sql_query(query_top_calories, conn)
top_calories
```

```python
## Visualization: Average Calories by Category

#We create a bar chart to visualize the average calorie values for each food category.

#Visualization helps us quickly understand which food groups tend to contain more calories.
```

```python
query = """
SELECT category, AVG(calories) as avg_calories
FROM foods
GROUP BY category
"""

result = pd.read_sql_query(query, conn)

plt.bar(result["category"], result["avg_calories"])
plt.xlabel("Category")
plt.ylabel("Average Calories")
plt.title("Average Calories by Category")
plt.show()
```

```python
query_high_protein = """
SELECT food_name, protein, calories
FROM foods
WHERE protein > 10
ORDER BY protein DESC
"""

high_protein = pd.read_sql_query(query_high_protein, conn)
high_protein
```

## Relationship Between Protein and Calories

The scatter plot shows the relationship between **protein content** and **calories**.

Each point represents a food item.

This visualization helps us see whether foods with higher protein tend to have higher calories.

```python
plt.figure(figsize=(8, 5))
plt.scatter(df["protein"], df["calories"])

for i in range(len(df)):
    plt.text(df["protein"][i], df["calories"][i], df["food_name"][i], fontsize=8)

plt.xlabel("Protein")
plt.ylabel("Calories")
plt.title("Protein vs Calories")
plt.tight_layout()
plt.show()
```
