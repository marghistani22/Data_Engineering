WHO Life Expectancy Data Engineering Project

```python
import requests
import pandas as pd
import sqlite3
import matplotlib.pyplot as plt
import os
```

```python
# Put your API key here if you have one
API_KEY = "YOUR_API_KEY_HERE"

# WHO API endpoint for life expectancy
API_URL = "https://ghoapi.azureedge.net/api/WHOSIS_000001"

# File names
RAW_FILE = "who_life_expectancy_raw.csv"
CLEAN_FILE = "who_life_expectancy_clean.csv"
DB_FILE = "who_life_expectancy.db"
```

```python
def extract_who_data(api_url, api_key=None):
    """
    Pipeline 1:
    Extract data from WHO API and return raw dataframe.
    """
    headers = {}

    # Optional API key
    if api_key and api_key != "YOUR_API_KEY_HERE":
        headers["Authorization"] = f"Bearer {api_key}"

    response = requests.get(api_url, headers=headers, timeout=60)
    response.raise_for_status()

    data = response.json()

    if "value" not in data:
        raise ValueError("No 'value' field found in API response.")

    raw_df = pd.DataFrame(data["value"])
    return raw_df
```

```python
raw_df = extract_who_data(API_URL, API_KEY)

print("Raw data shape:", raw_df.shape)
print(raw_df.head())

raw_df.to_csv(RAW_FILE, index=False)
print(f"Raw data saved to {RAW_FILE}")
```

```python
print("Columns in raw dataset:")
print(raw_df.columns.tolist())
```

```python
def transform_who_data(df):
    """
    Pipeline 2:
    Clean and transform WHO life expectancy data.
    """
    # Make a copy
    clean_df = df.copy()

    # Keep only useful columns if they exist
    selected_cols = [
        "SpatialDim",      # country code
        "ParentLocation",  # country / region
        "TimeDim",         # year
        "Dim1",            # sex code
        "NumericValue",    # numeric life expectancy
        "Value"            # text value
    ]

    available_cols = [col for col in selected_cols if col in clean_df.columns]
    clean_df = clean_df[available_cols]

    # Rename columns
    rename_dict = {
        "SpatialDim": "country_code",
        "ParentLocation": "country",
        "TimeDim": "year",
        "Dim1": "sex_code",
        "NumericValue": "life_expectancy",
        "Value": "value_text"
    }

    clean_df = clean_df.rename(columns=rename_dict)

    # Convert numeric fields
    if "year" in clean_df.columns:
        clean_df["year"] = pd.to_numeric(clean_df["year"], errors="coerce")

    if "life_expectancy" in clean_df.columns:
        clean_df["life_expectancy"] = pd.to_numeric(clean_df["life_expectancy"], errors="coerce")

    # If NumericValue is missing, try using Value
    if "life_expectancy" not in clean_df.columns and "value_text" in clean_df.columns:
        clean_df["life_expectancy"] = pd.to_numeric(clean_df["value_text"], errors="coerce")

    # Map sex codes to readable labels
    if "sex_code" in clean_df.columns:
        sex_map = {
            "MLE": "Male",
            "FMLE": "Female",
            "BTSX": "Both sexes"
        }
        clean_df["sex"] = clean_df["sex_code"].map(sex_map).fillna(clean_df["sex_code"])
    else:
        clean_df["sex"] = "Unknown"

    # Remove missing essential values
    required_cols = [col for col in ["country", "year", "life_expectancy"] if col in clean_df.columns]
    clean_df = clean_df.dropna(subset=required_cols)

    # Remove duplicates
    clean_df = clean_df.drop_duplicates()

    # Sort data
    sort_cols = [col for col in ["country", "year", "sex"] if col in clean_df.columns]
    clean_df = clean_df.sort_values(sort_cols).reset_index(drop=True)

    return clean_df
```

```python
clean_df = transform_who_data(raw_df)

print("Cleaned data shape:", clean_df.shape)
print(clean_df.head())

clean_df.to_csv(CLEAN_FILE, index=False)
print(f"Cleaned data saved to {CLEAN_FILE}")
```

```python
conn = sqlite3.connect(DB_FILE)

clean_df.to_sql("life_expectancy", conn, if_exists="replace", index=False)

print(f"Data loaded into SQLite database: {DB_FILE}")
```

```python
latest_year = int(clean_df["year"].max())
print("Latest year in dataset:", latest_year)

query_top_10 = f"""
SELECT country, AVG(life_expectancy) AS avg_life_expectancy
FROM life_expectancy
WHERE year = {latest_year}
AND sex = 'Both sexes'
GROUP BY country
ORDER BY avg_life_expectancy DESC
LIMIT 10
"""

top_10_countries = pd.read_sql_query(query_top_10, conn)
top_10_countries
```

```python
query_trend = """
SELECT year, AVG(life_expectancy) AS avg_life_expectancy
FROM life_expectancy
WHERE sex = 'Both sexes'
GROUP BY year
ORDER BY year
"""

trend_df = pd.read_sql_query(query_trend, conn)
trend_df.head()
```

```python
query_afghanistan = """
SELECT year, life_expectancy
FROM life_expectancy
WHERE country = 'Afghanistan'
AND sex = 'Both sexes'
ORDER BY year
"""

afghanistan_df = pd.read_sql_query(query_afghanistan, conn)
afghanistan_df.head()
```

```python
top_10_countries.to_csv("top_10_countries_latest_year.csv", index=False)
trend_df.to_csv("global_life_expectancy_trend.csv", index=False)
afghanistan_df.to_csv("afghanistan_life_expectancy.csv", index=False)

print("Reports exported successfully.")
```

```python
latest_year = clean_df["year"].max()

query = f"""
SELECT country, AVG(life_expectancy) as avg_life
FROM life_expectancy
WHERE year = {latest_year}
GROUP BY country
ORDER BY avg_life DESC
LIMIT 10
"""

top_10 = pd.read_sql_query(query, conn)

print(top_10)
```

```python
plt.figure(figsize=(10,6))

plt.bar(top_10["country"], top_10["avg_life"])

plt.xlabel("Country")
plt.ylabel("Life Expectancy")
plt.title(f"Top 10 Countries by Life Expectancy ({latest_year})")

plt.xticks(rotation=45)
plt.show()
```
