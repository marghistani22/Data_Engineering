```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans, DBSCAN
from sklearn.decomposition import PCA
```

## 1. Load the Dataset and Basic Cleaning

In this section, the dataset is loaded and cleaned.  
The column names are standardized, selected columns are converted into numeric format, missing values are handled, and the transaction date is converted into datetime format.



```python
df = pd.read_csv("bank_transactions_2025.csv")

df.columns = df.columns.str.strip()

numeric_fix_cols = ["TransactionAmount", "LoginAttempts", "CustomerAge", "AccountBalance"]
for c in numeric_fix_cols:
    if c in df.columns:
        df[c] = pd.to_numeric(df[c], errors="coerce")

if "LoginAttempts" in df.columns:
    df["LoginAttempts"] = df["LoginAttempts"].fillna(0).astype(int)

if "TransactionDate" in df.columns:
    df["TransactionDate"] = pd.to_datetime(df["TransactionDate"], errors="coerce")
```

 ## High-Value Transactions (Top 5%)

High-value transactions are defined as the top 5% of transaction amounts using the 95th percentile as a threshold. Transactions above this value are selected and counted by channel to identify where high-value activity occurs.

```python
# Step 1: Find threshold for top 5%
threshold = df["TransactionAmount"].quantile(0.95)

# Step 2: Filter high-value transactions
high_value = df[df["TransactionAmount"] >= threshold]

# Step 3: Count by channel
high_value_counts = high_value["Channel"].value_counts()

print(high_value_counts)
```

```python
plt.figure(figsize=(8, 5))

high_value_counts.sort_index().plot(kind="bar")

plt.title("High-Value Transactions (Top 5%) by Channel")
plt.xlabel("Channel")
plt.ylabel("Number of High-Value Transactions")

plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

#Statistical Analysis

### Transaction Amount

This section examines the distribution of transaction amounts and calculates the mean, median, standard deviation, and outliers using the Z-score method.

```python
amt = df["TransactionAmount"].dropna()

plt.figure(figsize=(8,5))
plt.hist(amt, bins=30)
plt.title("Distribution of Transaction Amount")
plt.xlabel("Transaction Amount")
plt.ylabel("Count")
plt.show()

mean_amt = amt.mean()
median_amt = amt.median()
std_amt = amt.std()

print("TransactionAmount Mean:", mean_amt)
print("TransactionAmount Median:", median_amt)
print("TransactionAmount Std Dev:", std_amt)

df["z_amount"] = (df["TransactionAmount"] - mean_amt) / std_amt
outliers_amt = df[df["z_amount"].abs() > 3]

print("TransactionAmount Outliers (|z| > 3):", len(outliers_amt))
print(outliers_amt[["TransactionAmount", "z_amount"]].head())
```

#### Transaction Duration

This section analyzes transaction duration using a histogram and descriptive statistics. Outliers are identified using the 3-standard-deviation rule.

```python
dur = df["TransactionDuration"].dropna()

plt.figure(figsize=(8,5))
plt.hist(dur, bins=30)
plt.title("Distribution of Transaction Duration")
plt.xlabel("Transaction Duration")
plt.ylabel("Count")
plt.show()

mean_dur = dur.mean()
median_dur = dur.median()
std_dur = dur.std()

print("TransactionDuration Mean:", mean_dur)
print("TransactionDuration Median:", median_dur)
print("TransactionDuration Std Dev:", std_dur)

df["z_duration"] = (df["TransactionDuration"] - mean_dur) / std_dur
outliers_dur = df[df["z_duration"].abs() > 3]

print("TransactionDuration Outliers (|z| > 3):", len(outliers_dur))
print(outliers_dur[["TransactionDuration", "z_duration"]].head())
```

### Account Balance

This section studies the distribution of account balance and reports the mean, median, standard deviation, and outliers.

```python
bal = df["AccountBalance"].dropna()

plt.figure(figsize=(8,5))
plt.hist(bal, bins=30)
plt.title("Distribution of Account Balance")
plt.xlabel("Account Balance")
plt.ylabel("Count")
plt.show()

mean_bal = bal.mean()
median_bal = bal.median()
std_bal = bal.std()

print("AccountBalance Mean:", mean_bal)
print("AccountBalance Median:", median_bal)
print("AccountBalance Std Dev:", std_bal)

df["z_balance"] = (df["AccountBalance"] - mean_bal) / std_bal
outliers_bal = df[df["z_balance"].abs() > 3]

print("AccountBalance Outliers (|z| > 3):", len(outliers_bal))
print(outliers_bal[["AccountBalance", "z_balance"]].head())
```

### Login Attempts

Since login attempts are discrete values, their distribution is shown using a discrete histogram. The mean, median, and standard deviation are also reported.

```python
login = df["LoginAttempts"].dropna().astype(int)

bins_login = np.arange(login.min() - 0.5, login.max() + 1.5, 1)

plt.figure(figsize=(8,5))
plt.hist(login, bins=bins_login, rwidth=0.9)
plt.title("Distribution of Login Attempts")
plt.xlabel("Login Attempts")
plt.ylabel("Count")
plt.xticks(range(login.min(), login.max() + 1))
plt.show()

mean_login = login.mean()
median_login = login.median()
std_login = login.std()

print("LoginAttempts Mean:", mean_login)
print("LoginAttempts Median:", median_login)
print("LoginAttempts Std Dev:", std_login)
```

### Daily Transaction Amount

To understand transaction behavior over time, the total transaction amount is aggregated by date. The distribution of daily transaction totals is then analyzed.

```python
df["TransactionDate"] = pd.to_datetime(df["TransactionDate"], errors="coerce")

daily_amount = (
    df.dropna(subset=["TransactionDate"])
      .groupby(df["TransactionDate"].dt.date)["TransactionAmount"]
      .sum()
)

plt.figure(figsize=(8,5))
plt.hist(daily_amount, bins=30)
plt.title("Distribution of Daily Transaction Amounts")
plt.xlabel("Total Transaction Amount per Day")
plt.ylabel("Number of Days")
plt.show()

mean_daily = daily_amount.mean()
median_daily = daily_amount.median()
std_daily = daily_amount.std()

print("Daily Mean:", mean_daily)
print("Daily Median:", median_daily)
print("Daily Std Dev:", std_daily)

z_daily = (daily_amount - mean_daily) / std_daily
daily_outliers = daily_amount[z_daily.abs() > 3]

print("Daily Outliers (|z| > 3):", len(daily_outliers))
print(daily_outliers.head())
```

## Risk Patterns in Fraud Detection

Fraud detection relies on identifying patterns in transaction data that differ from normal behavior. According to research, fraudulent activities can be detected by analyzing anomalies and unusual patterns within financial data. Machine learning techniques are particularly effective because they can learn complex relationships between variables and identify hidden fraud patterns. These patterns often involve combinations of features such as high transaction amounts, multiple login attempts, or unusual transaction channels. Therefore, detecting risk patterns is a key component of modern fraud detection systems.

### Age-Based Risk Pattern Analysis

Customer age can provide valuable insights into transaction behavior when combined with other variables. While age alone does not indicate fraud, unusual patterns such as very young or very old customers performing high-value transactions may represent potential risk.

For example, high transaction amounts associated with very young customers may be inconsistent with typical financial behavior, while elderly customers conducting large online transactions may be more vulnerable to fraud. Therefore, age is an important contextual factor in identifying suspicious transaction patterns.
