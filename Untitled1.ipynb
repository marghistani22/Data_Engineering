#Statistical Analysis

### Transaction Amount

This section examines the distribution of transaction amounts and calculates the mean, median, standard deviation, and outliers using the Z-score method.

```python
amt = df["TransactionAmount"].dropna()

plt.figure(figsize=(8,5))
plt.hist(amt, bins=30)
plt.title("Distribution of Transaction Amount")
plt.xlabel("Transaction Amount")
plt.ylabel("Count")
plt.show()

mean_amt = amt.mean()
median_amt = amt.median()
std_amt = amt.std()

print("TransactionAmount Mean:", mean_amt)
print("TransactionAmount Median:", median_amt)
print("TransactionAmount Std Dev:", std_amt)

df["z_amount"] = (df["TransactionAmount"] - mean_amt) / std_amt
outliers_amt = df[df["z_amount"].abs() > 3]

print("TransactionAmount Outliers (|z| > 3):", len(outliers_amt))
print(outliers_amt[["TransactionAmount", "z_amount"]].head())
```

#### Transaction Duration

This section analyzes transaction duration using a histogram and descriptive statistics. Outliers are identified using the 3-standard-deviation rule.

```python
dur = df["TransactionDuration"].dropna()

plt.figure(figsize=(8,5))
plt.hist(dur, bins=30)
plt.title("Distribution of Transaction Duration")
plt.xlabel("Transaction Duration")
plt.ylabel("Count")
plt.show()

mean_dur = dur.mean()
median_dur = dur.median()
std_dur = dur.std()

print("TransactionDuration Mean:", mean_dur)
print("TransactionDuration Median:", median_dur)
print("TransactionDuration Std Dev:", std_dur)

df["z_duration"] = (df["TransactionDuration"] - mean_dur) / std_dur
outliers_dur = df[df["z_duration"].abs() > 3]

print("TransactionDuration Outliers (|z| > 3):", len(outliers_dur))
print(outliers_dur[["TransactionDuration", "z_duration"]].head())
```

### Account Balance

This section studies the distribution of account balance and reports the mean, median, standard deviation, and outliers.

```python
bal = df["AccountBalance"].dropna()

plt.figure(figsize=(8,5))
plt.hist(bal, bins=30)
plt.title("Distribution of Account Balance")
plt.xlabel("Account Balance")
plt.ylabel("Count")
plt.show()

mean_bal = bal.mean()
median_bal = bal.median()
std_bal = bal.std()

print("AccountBalance Mean:", mean_bal)
print("AccountBalance Median:", median_bal)
print("AccountBalance Std Dev:", std_bal)

df["z_balance"] = (df["AccountBalance"] - mean_bal) / std_bal
outliers_bal = df[df["z_balance"].abs() > 3]

print("AccountBalance Outliers (|z| > 3):", len(outliers_bal))
print(outliers_bal[["AccountBalance", "z_balance"]].head())
```

### Daily Transaction Amount

To understand transaction behavior over time, the total transaction amount is aggregated by date. The distribution of daily transaction totals is then analyzed.

```python
df["TransactionDate"] = pd.to_datetime(df["TransactionDate"], errors="coerce")

daily_amount = (
    df.dropna(subset=["TransactionDate"])
      .groupby(df["TransactionDate"].dt.date)["TransactionAmount"]
      .sum()
)

plt.figure(figsize=(8,5))
plt.hist(daily_amount, bins=30)
plt.title("Distribution of Daily Transaction Amounts")
plt.xlabel("Total Transaction Amount per Day")
plt.ylabel("Number of Days")
plt.show()

mean_daily = daily_amount.mean()
median_daily = daily_amount.median()
std_daily = daily_amount.std()

print("Daily Mean:", mean_daily)
print("Daily Median:", median_daily)
print("Daily Std Dev:", std_daily)

z_daily = (daily_amount - mean_daily) / std_daily
daily_outliers = daily_amount[z_daily.abs() > 3]

print("Daily Outliers (|z| > 3):", len(daily_outliers))
print(daily_outliers.head())
```



### Pearson Correlation

Pearson correlation is a statistical method used to measure the linear relationship between two numerical variables. It shows how strongly and in what direction the variables are related.

The correlation value ranges from -1 to +1. A value close to +1 indicates a strong positive relationship, a value close to -1 indicates a strong negative relationship, and a value near 0 indicates no linear relationship.

```python
print("LoginAttempts vs TransactionAmount:",
      df["LoginAttempts"].corr(df["TransactionAmount"]))

print("TransactionDuration vs TransactionAmount:",
      df["TransactionDuration"].corr(df["TransactionAmount"]))

print("AccountBalance vs TransactionAmount:",
      df["AccountBalance"].corr(df["TransactionAmount"]))
```

Note: ### Why Correlation Was Used in This Project

Pearson correlation and the correlation matrix were used in this project to understand the relationships between numerical variables in the dataset. Since the data includes features such as Transaction Amount, Login Attempts, Transaction Duration, Customer Age, and Account Balance, it is important to analyze how these variables are related to each other.

Pearson correlation is suitable because it measures the linear relationship between two numerical variables, which helps identify whether variables increase or decrease together. This is useful in fraud detection, where suspicious behavior may involve combinations of variables, such as high transaction amounts with multiple login attempts.

The correlation matrix was used to examine all variable relationships at once, providing a complete overview of the dataset. This helps identify patterns, detect weak or strong relationships, and support further analysis such as clustering and anomaly detection.

Overall, correlation analysis helps reveal hidden patterns in the data and improves the understanding of transaction behavior in the context of fraud detection.

The correlation analysis was conducted to examine the relationships between key numerical variables in the dataset, including Transaction Amount, Login Attempts, Transaction Duration, Customer Age, and Account Balance. Overall, the results indicate that no single variable has a strong linear relationship with transaction amount. This highlights an important characteristic of fraud detection problems: fraudulent behavior is often complex and cannot be identified using a single feature. Instead, it requires analyzing combinations of multiple variables.

These findings suggest that correlation analysis alone is not sufficient for detecting fraud. Therefore, more advanced techniques such as clustering and anomaly detection are required to identify hidden patterns and unusual behaviors in the dataset.

RULE-BASED FLAG

```python
df["is_suspicious"] = (
    (df["TransactionAmount"] > 1000) &
    (df["LoginAttempts"] > 3) &
    (df["TransactionType"] == "Debit")
)

print("\n=== is_suspicious counts ===")
print(df["is_suspicious"].value_counts())

# Rule 1: Large debit with many login attempts
rule1 = (
    (df["TransactionAmount"] > 1000) &
    (df["LoginAttempts"] > 3) &
    (df["TransactionType"] == "Debit")
)

# Rule 2: Extremely large transaction
rule2 = df["TransactionAmount"] > 2000

# Rule 3: Many login attempts
rule3 = df["LoginAttempts"] >= 4

# Combine rules
df["is_suspicious"] = rule1 | rule2 | rule3

print(df["is_suspicious"].value_counts())
```

### Rule-Based Fraud Detection Results

A rule-based approach was applied to identify potentially suspicious transactions. The rules were designed based on domain knowledge, including high transaction amounts, multiple login attempts, and debit transactions.

The results show that only a small number of transactions were flagged as suspicious compared to the total number of transactions. This indicates that the rules are strict and capture only extreme cases of potentially fraudulent behavior.

Although rule-based methods are simple and easy to interpret, they may fail to detect more complex or hidden fraud patterns. Therefore, additional techniques such as clustering and anomaly detection are required to identify subtle and non-obvious suspicious activities.

### Clustering

The K-Means algorithm is a simple algorithm capable of clustering this kind of dataset very quickly and efficiently, often in just a few iterations. In K-Means clustering, each data point is usually assigned to a single cluster,

In this study, features such as Transaction Amount, Login Attempts, Customer Age, and Account Balance were used. The goal is to identify groups of normal and potentially abnormal transactions.


```python
cluster_features = ["TransactionAmount", "LoginAttempts", "CustomerAge", "AccountBalance"]
X_cluster = df[cluster_features].dropna()

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X_cluster)

kmeans = KMeans(n_clusters=3, random_state=42, n_init=10)
df.loc[X_cluster.index, "kmeans_cluster"] = kmeans.fit_predict(X_scaled)

print("\n  K-Means cluster counts ")
print(df["kmeans_cluster"].value_counts())
```

```python


# Select features for clustering
features = ['TransactionAmount', 'TransactionDuration']
X = df[features].copy()

# Standardize the data
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Determine the optimal number of clusters using the Elbow Method
inertia = []
K = range(1, 10)  # Test clusters from 1 to 9

for k in K:
    kmeans = KMeans(n_clusters=k, random_state=0, n_init=10)
    kmeans.fit(X_scaled)
    inertia.append(kmeans.inertia_)

# Plot the Elbow Method
plt.figure()
plt.plot(K, inertia, marker='o')
plt.xlabel("Number of Clusters (k)")
plt.ylabel("Inertia")
plt.title("Elbow Method for Optimal Number of Clusters")
plt.show()
```
